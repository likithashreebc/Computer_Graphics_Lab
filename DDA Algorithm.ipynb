```python
import matplotlib.pyplot as plt


def dda_line(x1, y1, x2, y2):
    points = []

    dx = x2 - x1
    dy = y2 - y1

    steps = max(abs(dx), abs(dy))

    x_increment = dx / steps
    y_increment = dy / steps

    x = x1
    y = y1

    for i in range(steps + 1):
        points.append((round(x), round(y)))

        x = x + x_increment
        y = y + y_increment

    return points


# Input Point A
x1 = int(input("Enter x1 of point A: "))
y1 = int(input("Enter y1 of point A: "))

# Input Point B
x2 = int(input("Enter x2 of point B: "))
y2 = int(input("Enter y2 of point B: "))

# Input colour
line_color = input("Enter line colour: ")


# Calculate DDA points
points = dda_line(x1, y1, x2, y2)


# Display calculated points
print("\nPoints calculated using DDA Algorithm:")

for point in points:
    print(point)


# Extract coordinates
x_points = [point[0] for point in points]
y_points = [point[1] for point in points]


# Draw line
plt.figure(figsize=(8, 6))

plt.plot(
    x_points,
    y_points,
    'o-',
    color=line_color,
    linewidth=2
)

# Mark A and B
plt.scatter(
    [x1, x2],
    [y1, y2],
    color='black',
    s=80
)

plt.text(x1, y1, "  A")
plt.text(x2, y2, "  B")

plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("DDA Line Drawing Algorithm")

plt.grid(True)

plt.xticks(range(min(x1, x2) - 1, max(x1, x2) + 2))
plt.yticks(range(min(y1, y2) - 1, max(y1, y2) + 2))

plt.show()
```

```python
import matplotlib.pyplot as plt
import math


def dda_circle(xc, yc, r):
    points = []

    theta = 0
    theta_increment = 1 / r

    while theta <= math.pi / 4:
        x = round(r * math.cos(theta))
        y = round(r * math.sin(theta))

        # 8-way symmetry
        points.append((xc + x, yc + y))
        points.append((xc - x, yc + y))
        points.append((xc + x, yc - y))
        points.append((xc - x, yc - y))

        points.append((xc + y, yc + x))
        points.append((xc - y, yc + x))
        points.append((xc + y, yc - x))
        points.append((xc - y, yc - x))

        theta += theta_increment

    return list(set(points))


# -------------------------------
# INPUT
# -------------------------------

xc = int(input("Enter x-coordinate of center: "))
yc = int(input("Enter y-coordinate of center: "))

r = int(input("Enter radius: "))

circle_color = input(
    "Enter circle colour (red/blue/green/black/orange/purple): "
).strip().lower()


# -------------------------------
# VALIDATE COLOUR
# -------------------------------

valid_colors = [
    "red",
    "blue",
    "green",
    "black",
    "orange",
    "purple",
    "pink",
    "yellow",
    "brown",
    "gray",
    "cyan",
    "magenta"
]

if circle_color not in valid_colors:
    print("Invalid colour! Using blue instead.")
    circle_color = "blue"


# -------------------------------
# CALCULATE POINTS
# -------------------------------

points = dda_circle(xc, yc, r)


print("\nPoints calculated using DDA Circle Algorithm:")

for point in sorted(points):
    print(point)


# -------------------------------
# EXTRACT X AND Y
# -------------------------------

x_points = [point[0] for point in points]
y_points = [point[1] for point in points]


# -------------------------------
# DRAW CIRCLE
# -------------------------------

plt.figure(figsize=(8, 8))

plt.scatter(
    x_points,
    y_points,
    color=circle_color,
    s=30
)

# Draw center
plt.scatter(
    xc,
    yc,
    color="black",
    s=80
)

plt.text(
    xc,
    yc,
    "  Center",
    fontsize=12
)

plt.xlabel("X-axis")
plt.ylabel("Y-axis")

plt.title("DDA Circle Drawing Algorithm")

plt.grid(True)
plt.axis("equal")

plt.show()
```
