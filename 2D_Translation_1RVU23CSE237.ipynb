2D Translation

```python
import numpy as np
import matplotlib.pyplot as plt

# Enter number of points
n = int(input("Enter number of points: "))

points = []

for i in range(n):
    x, y = map(float, input(f"Enter point {i+1} (x y): ").split())
    points.append([x, y, 1])

points = np.array(points).T

# Enter translation values
tx = float(input("Enter translation in X (tx): "))
ty = float(input("Enter translation in Y (ty): "))

# Translation matrix
T = np.array([
    [1, 0, tx],
    [0, 1, ty],
    [0, 0, 1]
])

# Apply transformation
transformed = T @ points

# Display coordinates
print("\nOriginal Points:")
print(points[:2].T)

print("\nTranslated Points:")
print(transformed[:2].T)

# Plot
plt.figure(figsize=(7, 6))

plt.plot(points[0], points[1], 'bo-', label="Original")
plt.plot(transformed[0], transformed[1], 'ro-', label="Translated")

plt.axhline(0, color='black')
plt.axvline(0, color='black')
plt.grid()
plt.xlabel("X")
plt.ylabel("Y")
plt.title("2D Translation using Homogeneous Coordinates")
plt.legend()
plt.axis("equal")
plt.show()
```

2. Rotation

```python
import numpy as np
import matplotlib.pyplot as plt

# Enter number of points
n = int(input("Enter number of points: "))

points = []

for i in range(n):
    x, y = map(float, input(f"Enter point {i+1} (x y): ").split())
    points.append([x, y, 1])

points = np.array(points).T

# Enter rotation angle
angle = float(input("Enter rotation angle in degrees: "))

theta = np.radians(angle)

# Rotation matrix
R = np.array([
    [np.cos(theta), -np.sin(theta), 0],
    [np.sin(theta),  np.cos(theta), 0],
    [0, 0, 1]
])

# Apply transformation
transformed = R @ points

# Display coordinates
print("\nOriginal Points:")
print(points[:2].T)

print("\nRotated Points:")
print(transformed[:2].T)

# Plot
plt.figure(figsize=(7, 6))

plt.plot(points[0], points[1], 'bo-', label="Original")
plt.plot(transformed[0], transformed[1], 'ro-', label="Rotated")

plt.axhline(0, color='black')
plt.axvline(0, color='black')
plt.grid()
plt.xlabel("X")
plt.ylabel("Y")
plt.title("2D Rotation using Homogeneous Coordinates")
plt.legend()
plt.axis("equal")
plt.show()
```

3. Reflection

```python
import numpy as np
import matplotlib.pyplot as plt

# Enter number of points
n = int(input("Enter number of points: "))

points = []

for i in range(n):
    x, y = map(float, input(f"Enter point {i+1} (x y): ").split())
    points.append([x, y, 1])

points = np.array(points).T

print("\nChoose Reflection:")
print("1. Reflection about X-axis")
print("2. Reflection about Y-axis")
print("3. Reflection about Origin")

choice = int(input("Enter your choice: "))

if choice == 1:
    # Reflection about X-axis
    F = np.array([
        [1, 0, 0],
        [0, -1, 0],
        [0, 0, 1]
    ])
    title = "Reflection about X-axis"

elif choice == 2:
    # Reflection about Y-axis
    F = np.array([
        [-1, 0, 0],
        [0, 1, 0],
        [0, 0, 1]
    ])
    title = "Reflection about Y-axis"

elif choice == 3:
    # Reflection about Origin
    F = np.array([
        [-1, 0, 0],
        [0, -1, 0],
        [0, 0, 1]
    ])
    title = "Reflection about Origin"

else:
    print("Invalid choice")
    exit()

# Apply transformation
transformed = F @ points

# Display coordinates
print("\nOriginal Points:")
print(points[:2].T)

print("\nReflected Points:")
print(transformed[:2].T)

# Plot
plt.figure(figsize=(7, 6))

plt.plot(points[0], points[1], 'bo-', label="Original")
plt.plot(transformed[0], transformed[1], 'ro-', label="Reflected")

plt.axhline(0, color='black')
plt.axvline(0, color='black')
plt.grid()
plt.xlabel("X")
plt.ylabel("Y")
plt.title(title)
plt.legend()
plt.axis("equal")
plt.show()
```

4. Scaling

```python
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------------------
# Enter number of points
# -----------------------------------------
n = int(input("Enter number of points: "))

points = []

for i in range(n):
    x, y = map(float, input(f"Enter point {i+1} (x y): ").split())
    points.append([x, y, 1])

points = np.array(points).T


# -----------------------------------------
# Enter scaling factors
# -----------------------------------------
sx = float(input("Enter scaling factor in X (sx): "))
sy = float(input("Enter scaling factor in Y (sy): "))


# -----------------------------------------
# Choose scaling type
# -----------------------------------------
print("\nChoose scaling:")
print("1. Scaling about Origin")
print("2. Scaling about Fixed Point")

choice = int(input("Enter your choice: "))


# -----------------------------------------
# Scaling about Origin
# -----------------------------------------
if choice == 1:

    S = np.array([
        [sx, 0,  0],
        [0,  sy, 0],
        [0,  0,  1]
    ])

    transformed = S @ points

    title = "Scaling about Origin"


# -----------------------------------------
# Scaling about Fixed Point
# -----------------------------------------
elif choice == 2:

    xf = float(input("Enter fixed point X: "))
    yf = float(input("Enter fixed point Y: "))

    # Translate fixed point to origin
    T1 = np.array([
        [1, 0, -xf],
        [0, 1, -yf],
        [0, 0, 1]
    ])

    # Scaling matrix
    S = np.array([
        [sx, 0,  0],
        [0,  sy, 0],
        [0,  0,  1]
    ])

    # Translate back
    T2 = np.array([
        [1, 0, xf],
        [0, 1, yf],
        [0, 0, 1]
    ])

    # Combined transformation
    M = T2 @ S @ T1

    transformed = M @ points

    title = "Scaling about Fixed Point"


else:
    print("Invalid choice")
    exit()


# -----------------------------------------
# Display coordinates
# -----------------------------------------
print("\nOriginal Points:")
print(points[:2].T)

print("\nTransformed Points:")
print(np.round(transformed[:2].T, 2))


# -----------------------------------------
# Plot
# -----------------------------------------
plt.figure(figsize=(7, 6))

plt.plot(
    points[0],
    points[1],
    'bo-',
    label="Original"
)

plt.plot(
    transformed[0],
    transformed[1],
    'ro-',
    label="Transformed"
)

# Show fixed point
if choice == 2:
    plt.plot(
        xf,
        yf,
        'go',
        markersize=10,
        label="Fixed Point"
    )

plt.axhline(0, color='black')
plt.axvline(0, color='black')

plt.grid(True)
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title(title)
plt.legend()
plt.axis("equal")

plt.show()
```
